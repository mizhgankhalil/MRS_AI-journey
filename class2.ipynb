```python
#split the bill
total_bill = int(input("total bill:"))
friends = int(input("number of friends:"))
share = total_bill // friends
remaining = total_bill % friends
print(f"each friend will pay: Rs.{share}")
print(f"remaining amount: Rs.{remaining}")
```
